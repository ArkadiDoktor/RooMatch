```python
import os
import re
import csv
import random
from pathlib import Path
from typing import Dict, List

import torch
from PIL import Image, ImageDraw
from diffusers import StableDiffusionInpaintPipeline
```

```python
ROOM_ATTRIBUTES = {
    "furniture": [
        "replace the sofa with a different style sofa",
        "add a small coffee table in front of the sofa",
        "remove one piece of furniture from the room"
    ],
    "decor": [
        "add decorative pillows and throws",
        "add framed artwork on the walls without changing wall positions",
        "add small decorative objects on shelves and tables"
    ],
    "plants": [
        "add one medium indoor plant in a corner",
        "add several small potted plants on surfaces",
        "remove all indoor plants"
    ],
    "people": [
        "add one person sitting casually on the sofa",
        "add a person standing naturally in the room"
    ],
    "clutter": [
        "make the room very tidy and minimal",
        "add light everyday clutter like books and magazines",
        "slightly rearrange small objects on surfaces"
    ],
    "materials": [
        "change fabric materials of furniture only",
        "change color of pillows and textiles",
        "change furniture upholstery while keeping shapes identical"
    ]
}

```

```python
BASE_PROMPT = (
    "Photorealistic INPAINTING of the SAME physical room as the input image. "
    "This is an interior edit, NOT a new image. "
    "Keep EXACT room structure: same walls, same wall positions, "
    "same windows, same doors, same ceiling, same floor, same camera position. "
    "Do NOT add, remove, resize, move, or invent windows, doors, openings, or walls. "
    "Only modify movable interior objects such as furniture, decor, people, plants, "
    "and small items. "
    "The room layout and geometry must remain IDENTICAL to the input image."
)


NEGATIVE_PROMPT = (
    "new room, different room, room redesign, architectural changes, "
    "changed layout, changed geometry, perspective change, camera move, "
    "added window, removed window, extra window, fake window, "
    "added door, removed door, extra door, fake door, "
    "wall opening, doorway, archway, hallway, "
    "structural changes, walls moved, walls removed, "
    "ceiling changed, floor changed, "
    "distorted walls, warped geometry, unrealistic perspective, "
    "low quality, blurry, artifacts"
)

```

```python
def random_inpaint_mask(w: int, h: int) -> Image.Image:
    mask = Image.new("L", (w, h), 0)
    draw = ImageDraw.Draw(mask)

    # large central region
    draw.rectangle(
        [int(w*0.1), int(h*0.1), int(w*0.9), int(h*0.9)],
        fill=255
    )

    # decor-heavy lower regions
    for _ in range(3):
        x0 = random.randint(0, w//2)
        y0 = random.randint(h//3, h-1)
        x1 = x0 + random.randint(w//4, w//2)
        y1 = y0 + random.randint(h//6, h//3)
        draw.rectangle([x0, y0, x1, y1], fill=255)

    return mask.resize((w//2, h//2), Image.BILINEAR).resize((w, h), Image.BILINEAR)
```

```python
def generate_attributed_prompt(min_attrs=2, max_attrs=4) -> Dict:
    keys = random.sample(
        list(ROOM_ATTRIBUTES.keys()),
        random.randint(min_attrs, max_attrs)
    )

    actions = [random.choice(ROOM_ATTRIBUTES[k]) for k in keys]

    prompt = (
        BASE_PROMPT
        + " Apply the following changes: "
        + "; ".join(actions)
        + ". High realism, straight lines."
    )

    return {
        "prompt": prompt,
        "negative_prompt": NEGATIVE_PROMPT,
        "attributes": dict(zip(keys, actions))
    }
```

```python
@torch.inference_mode()
def generate_room_variants(
    pipe,
    image: Image.Image,
    num_variants: int = 10,
    seed: int = 0,
    size: int = 512
) -> List[Dict]:

    image = image.convert("RGB").resize((size, size), Image.LANCZOS)
    w, h = image.size

    results = []
    generator = torch.Generator(device=pipe.device)

    for i in range(num_variants):
        meta = generate_attributed_prompt()
        mask = random_inpaint_mask(w, h)

        generator.manual_seed(seed + i)

        out = pipe(
            prompt=meta["prompt"],
            negative_prompt=meta["negative_prompt"],
            image=image,
            mask_image=mask,
            guidance_scale=6.5,
            num_inference_steps=35,
            strength=0.65,
            generator=generator,
        ).images[0]

        results.append({
            "image": out,
            "attributes": meta["attributes"],
            "seed": seed + i
        })

    return results
```

```python
def build_dataset(
    input_dir: str,
    output_dir: str,
    num_variants: int = 10
):
    IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".bmp"}

    def extract_room_id(path: Path, fallback_id: int) -> int:
        match = re.search(r"(\d+)", path.stem)
        return int(match.group(1)) if match else fallback_id

    device = "cuda" if torch.cuda.is_available() else "cpu"

    pipe = StableDiffusionInpaintPipeline.from_pretrained( "runwayml/stable-diffusion-inpainting",
    torch_dtype=torch.float16 if device == "cuda" else torch.float32, safety_checker=None
    ).to(device)

    pipe.enable_attention_slicing()

    input_dir = Path(input_dir)
    output_dir = Path(output_dir)
    real_dir = output_dir / "real"
    synth_dir = output_dir / "synthetic"

    real_dir.mkdir(parents=True, exist_ok=True)
    synth_dir.mkdir(parents=True, exist_ok=True)

    rows = []

    image_paths = sorted(
        p for p in input_dir.iterdir()
        if p.suffix.lower() in IMAGE_EXTENSIONS
    )

    for idx, img_path in enumerate(image_paths):
        room_id = extract_room_id(img_path, idx)
        print(f"Processing room {room_id} ({img_path.name})")

        image = Image.open(img_path).convert("RGB")
        real_path = real_dir / img_path.name
        image.save(real_path)

        rows.append({
            "path": str(real_path),
            "room_id": room_id,
            "is_synthetic": 0
        })

        variants = generate_room_variants(
            pipe=pipe,
            image=image,
            num_variants=num_variants,
            seed=room_id * 1000
        )

        room_out = synth_dir / f"room_{room_id}"
        room_out.mkdir(parents=True, exist_ok=True)

        for i, v in enumerate(variants):
            out_path = room_out / f"syn_{i:02d}.png"
            v["image"].save(out_path)
            rows.append({
                "path": str(out_path),
                "room_id": room_id,
                "is_synthetic": 1
            })

    with open(output_dir / "labels.csv", "w", newline="") as f:
        writer = csv.DictWriter(
            f,
            fieldnames=["path", "room_id", "is_synthetic"]
        )
        writer.writeheader()
        writer.writerows(rows)

    print("\nDataset generation complete.")
```

```python
build_dataset(
    input_dir="/content/input_rooms",
    output_dir="/content/dataset",
    num_variants=10
)
```
